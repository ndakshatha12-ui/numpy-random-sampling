# NumPy Random Sampling

```python
import numpy as np

rng = np.random.default_rng(42)
```

## Random integers and floating-point numbers

```python
# random integers between 1 and 100
ints = rng.integers(1, 100, size=10)
print(ints)
```

```python
# random floats between 0 and 1
floats = rng.random(5)
print(floats)
```

```python
# random floats scaled to a custom range
scaled = rng.uniform(10, 20, size=5)
print(scaled)
```

```python
# random numbers from a normal distribution
normal_vals = rng.normal(loc=50, scale=10, size=5)
print(normal_vals)
```

## Random samples from an array

```python
students = np.array(["Asha", "Raj", "Meena", "Vikram", "Priya", "Karan"])

# sample without replacement - no repeats
sample = rng.choice(students, size=3, replace=False)
print(sample)
```

```python
# sample with replacement - repeats allowed
sample_wr = rng.choice(students, size=4, replace=True)
print(sample_wr)
```

```python
# shuffling an array in place
deck = np.arange(1, 11)
rng.shuffle(deck)
print(deck)
```

## Reproducibility using a seed

```python
# same seed -> identical results
rng1 = np.random.default_rng(7)
rng2 = np.random.default_rng(7)

a = rng1.integers(1, 100, size=5)
b = rng2.integers(1, 100, size=5)

print(a)
print(b)
print("Identical:", np.array_equal(a, b))
```

```python
# no fixed seed -> different results each time
rng3 = np.random.default_rng()
rng4 = np.random.default_rng()

c = rng3.integers(1, 100, size=5)
d = rng4.integers(1, 100, size=5)

print(c)
print(d)
print("Identical:", np.array_equal(c, d))
```

## Generating a small simulated dataset

```python
# 6 students, 3 subjects, marks between 40 and 100
sim_marks = rng.integers(40, 100, size=(6, 3))
print(sim_marks)
```

```python
print("Average per subject:", np.mean(sim_marks, axis=0))
```

## Interview questions

**Why are random numbers useful in data science?**
They're used to simulate data when real data isn't available yet, to split datasets into train/test sets, to shuffle data so patterns aren't learned from ordering, and to initialize models before training. Random sampling also underlies techniques like bootstrapping and Monte Carlo simulation for estimating uncertainty.

**What is a random seed?**
A starting value that initializes the random number generator's internal state. The sequence of "random" numbers a computer produces is actually deterministic given that starting point, so setting the same seed makes the generator produce the exact same sequence every time it runs.

**Why is reproducibility important?**
Without a fixed seed, results change every run, which makes debugging, testing, and comparing experiments unreliable — a change in output could be from a code fix or just from a different random draw. Setting a seed means results can be checked, shared, and reproduced exactly by someone else running the same code.
